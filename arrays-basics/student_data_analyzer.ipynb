```python
student = {
    "name": "Dev",
    "age": 36,
    "course": "AI_ML",
    "marks": [65, 70, 80, 75, 85]
}

print("Student Name:", student["name"])
print("Course:", student["course"])

total = 0
for m in student["marks"]:
    total += m

average = total / len(student["marks"])
print("Average Marks:", average)
```
