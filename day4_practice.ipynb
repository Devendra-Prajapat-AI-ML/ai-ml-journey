```python
import pandas as pd
df = pd.read_csv('titanic.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
# उम्र की खाली जगहों को औसत (Mean) से भरना
df['Age'] = df['Age'].fillna(df['Age'].mean())

# अब दोबारा चेक करते हैं कि क्या अब भी कोई खाली जगह बची है?
print(df['Age'].isnull().sum())
```

```python
# now we have to remove the column cabin as it is not needed and it is approx empty.
df.drop('Cabin', axis=1, inplace=True)

# Check karne ke liye ki kya ab bhi wo column hai?
print(df.columns)
```

```python
# 'Sex' कॉलम को नंबर में बदलना 
df['Sex'] = df['Sex'].map({'male': 1, 'female': 0})

# चेक करने के लिए पहली 5 लाइनें देखें
df.head()
```
